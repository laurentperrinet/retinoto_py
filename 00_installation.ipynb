6. download models 

`````python
# ----------------------------------------------------------------------
import os
import torchvision.models as models
# ----------------------------------------------------------------------
# Tell torchvision to use that folder
os.environ.setdefault('TORCH_HOME', str(args.data_cache))
models_to_download = [
    lambda: models.resnet18(weights=models.ResNet18_Weights.DEFAULT),
    lambda: models.resnet50(weights=models.ResNet50_Weights.DEFAULT),
    lambda: models.resnet101(weights=models.ResNet101_Weights.DEFAULT),
    lambda: models.convnext_tiny(weights=models.ConvNeXt_Tiny_Weights.IMAGENET1K_V1),
    lambda: models.convnext_small(weights=models.ConvNeXt_Small_Weights.IMAGENET1K_V1),
    lambda: models.convnext_base(weights=models.ConvNeXt_Base_Weights.IMAGENET1K_V1),
    lambda: models.convnext_large(weights=models.ConvNeXt_Large_Weights.IMAGENET1K_V1),
]

for factory in models_to_download:
    factory()            # triggers download into data_cache
print("[✔] All pretrained models cached in", args.data_cache)
`````

Note that we will use the localisation data from https://www.kaggle.com/c/imagenet-object-localization-challenge/data
